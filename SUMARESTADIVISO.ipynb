```python
numero_a = 1500
numero_b = 250

resultado_suma = numero_a + numero_b
print(f"Suma:= {resultado_suma}")

resultado_resta = numero_a - numero_b
print(f"Resta:{resultado_resta}")

resultado_division = numero_a / numero_b
print(f"Resta:{resultado_division}")
```
