SISTEMA DE LOOG BANK SALDO INGRESADO 1000; SUPERA 1000 se pierde

```python
def simular_caja():
    limite_caja = 1000
    total_retirado = 0

    print(f"--- Simulador de Caja (Límite máximo de retiro acumulado: {limite_caja} pesos) ---")

    while True:
        saldo_disponible = limite_caja - total_retirado
        print(f"\nSaldo disponible para retirar: {saldo_disponible} pesos")

        try:
            monto_retiro = float(input("Ingrese la cantidad a retirar (o 0 para salir): "))
        except ValueError:
            print("Por favor, ingrese un número válido.")
            continue

        if monto_retiro == 0:
            print("Operación finalizada por el usuario.")
            break

        if monto_retiro < 0:
            print("No puede retirar cantidades negativas.")
            continue
                    if total_retirado + monto_retiro > limite_caja:
            print(f"\n[ALERTA] El retiro de {monto_retiro} pesos supera el límite de {limite_caja} pesos.")
            print("Se ha superado el límite permitido. Cerrando el programa (break).")
            break

        total_retirado += monto_retiro
        print(f"Retiro exitoso de {monto_retiro} pesos. Total retirado: {total_retirado} pesos.")

        if total_retirado == limite_caja:
            print("Ha alcanzado exactamente el límite de 1000 pesos. Cerrando caja.")
            break
simular_caja()
```
