**Una fabrica necesita crear 100 piezas. Realiza un programa que solicita cuantas piezas se fabricaron en cuanto lote y se acomule. Y el programa debera seguir solicitando datos hasta alcanzar o superar la meta**


1.   Iniciar el total de piezas en 0
2.   Solicitar la cantidad de piezas en el siguiente lote

3. Sumar esa cantidad al total y mostrarlo en el avance
4. Repetir el proceso mientras el total sea menor de 100
5. Mostrar un mensaje cuando se alcance la meta

```python
# 1. Iniciar el total de piezas en 0
total_piezas = 0
meta = 100

# 4. Repetir el proceso mientras el total sea menor de 100
while total_piezas < meta:
    try:
        # 2. Solicitar la cantidad de piezas en el siguiente lote
        lote = int(input(f"Ingrese la cantidad de piezas del siguiente lote (Faltan {meta - total_piezas} piezas): "))

        if lote < 0:
            print("Por favor, ingrese un número positivo.")
            continue

        # 3. Sumar esa cantidad al total y mostrarlo en el avance
        total_piezas += lote
        print(f"Avance actual: {total_piezas} / {meta} piezas.\n")
    except ValueError:
        print("Entrada no válida. Por favor, ingrese un número entero.")

# 5. Mostrar un mensaje cuando se alcance la meta
print(f"¡Felicidades! Se ha alcanzado o superado la meta con un total de {total_piezas} piezas.")
```
