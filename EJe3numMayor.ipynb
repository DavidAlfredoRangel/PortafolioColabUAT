Escriba un programa que lea 3 números  y determine cual es el mayor.

```python
#Descripción en lenguaje natural : Primero se mandaran los numeros agregados al if, else
#para determinar cual es el mayor y menor

def determinar_mayor(a, b, c):
  #
    if a > b:
        if a > c:
            return a
        else:
            return c
    else:
        if b > c:
            return b
        else:
            return c

# Aqui pediremos los 3 numeros con el input
try:
    num1 = int(input("Ingrese el primer número: "))
    num2 = int(input("Ingrese el segundo número: "))
    num3 = int(input("Ingrese el tercer número: "))

    mayor = determinar_mayor(num1, num2, num3)
    print(f"El mayor de los tres números es: {mayor}")
except ValueError:
    print("Por favor, ingrese valores numéricos válidos.")
```

Diagrama de flujo


```python
#
from graphviz import Digraph

# Crear un nuevo diagrama de flujo para encontrar el mayor de 3 números
dot = Digraph(comment='Mayor de 3 números')

# Definir nodos
dot.node('A', 'Inicio', shape='oval')
dot.node('B', 'Leer A, B, C', shape='parallelogram')
dot.node('C', '¿A > B?', shape='diamond')
dot.node('D', '¿A > C?', shape='diamond')
dot.node('E', '¿B > C?', shape='diamond')
dot.node('F', 'El mayor es A', shape='box')
dot.node('G', 'El mayor es C', shape='box')
dot.node('H', 'El mayor es B', shape='box')
dot.node('I', 'Fin', shape='oval')

# Definir conexiones (flechas)
dot.edge('A', 'B')
dot.edge('B', 'C')

# Ramas de la primera condición (¿A > B?)
dot.edge('C', 'D', label='Sí')
dot.edge('C', 'E', label='No')

# Ramas de la condición (¿A > C?)
dot.edge('D', 'F', label='Sí')
dot.edge('D', 'G', label='No')

# Ramas de la condición (¿B > C?)
dot.edge('E', 'H', label='Sí')
dot.edge('E', 'G', label='No')

# Conexiones finales a Fin
dot.edge('F', 'I')
dot.edge('G', 'I')
dot.edge('H', 'I')

# Mostrar el diagrama
dot
```
