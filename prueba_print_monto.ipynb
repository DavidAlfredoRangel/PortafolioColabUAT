PROGRAMA QUE PIDE INGRESAR VALOR Y RESTA EL IVA



```python

try:
    # Se solicita al usuario ingresar el monto total (con IVA incluido)
    monto_total = float(input("Ingresa la cantidad de dinero en pesos ($): "))

    # En México, el IVA estándar es del 16% (0.16)
    # Para obtener el valor sin IVA (subtotal), dividimos el total entre 1.16
    monto_sin_iva = monto_total / 1.16
    iva_calculado = monto_total - monto_sin_iva

    # Se muestran los resultados en pantalla con formato de dos decimales
    print(f"\n--- Desglose de IVA (16% México) ---")
    print(f"Monto total ingresado: ${monto_total:,.2f} MXN")
    print(f"Monto neto (sin IVA):  ${monto_sin_iva:,.2f} MXN")
    print(f"IVA de la cantidad:    ${iva_calculado:,.2f} MXN")

except ValueError:
    print("Error: Por favor, introduce un número válido para el dinero.")
```
