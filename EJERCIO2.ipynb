**Programa para calcular el precio final de un producto que tiene un descuento. (Solicitar el % del Descuento)**

Codigo final

```python
precioI = float(input("Ingresa el monto inicial del producto : "))
descuento = int(input("Porfavor ingresar al porcentaje de descuento del producto : "))


precio2 = precioI * (descuento / 100)

final = precioI - precio2

print(f"El porcentaje del descuento es {descuento}% y el monto a pagar del producto es de: {final}")
```

A continuacion exxplicare paso por paso desglozando el codigo

```python
#Lo primero que veremos es que se creara una variable en el mismo input para pedir el precio inicial del producto
precioI = float(input("Ingresa el monto inicial del producto : "))

#Posteriormente volveremos a crear otra variable con un nombre diferente para poder pedir el descuento
descuento = int(input("Porfavor ingresar al porcentaje de descuento del producto : "))


#A continuacion se hara una operacion donde se sacara la cantidad de dinero del descuento
#Lo que se hara es dividir el porcentaje que se ingreso entre 100% para luego multiplicarlo por el monto inical
precio2 = precioI * (descuento / 100)
#Ejemplo: 50 = 500 *(10% / 100)


#Y despues seguiria una ultima operacion para sacar el monto a pagar en total
#Es restar el monto inicial con el descuento que se saco anteriormente
final = precioI - precio2
#Ejemplo 450 = 500 - 50


#Para finalizar se usara un print para mostrar el resultado en la pantalla de la conzola

print(f"El porcentaje del descuento es {descuento}% y el monto a pagar del producto es de: {final}")
```

# **Programa para convertir Radianes a grados. **

```python
import math

radianes = float(input("Ingresa los radianes : "))

grados = radianes * 180/math.pi
print(f"Los radianes convertidos a grados son {grados}°")
```

```python
#La instrucción import math sirve para cargar el módulo matemático integrado de Python
#lo que te permite usar funciones y constantes numéricas avanzadas sin tener que programarlas desde cero
import math


#Entonces tenemos que pedir los radianes en el input
radianes = float(input("Ingresa los radianes : "))



# Para no tener que escribir pi usamos el valor exacto de Pi que maneja Python que es
#math.pi     Resultado: 3.141592653589793
grados = radianes * 180/math.pi


#Y por ultimo hacemos un print para imprimir el resultado de la operacion
print(f"Los radianes convertidos a grados son {grados}°")
#
```
