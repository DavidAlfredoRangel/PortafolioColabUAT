```python
def calcular_mcd(a, b):
    # AQui primero se hara la operacion del mdc
    while b != 0:
        residuo = a % b
        a = b
        b = residuo
    return a

# Solicitar que se ingresen los dos números para poder sacar el MCD
try:

    numero1 = int(input("Ingrese el primer número que sea entero: ") )
    numero2 = int(input("Ingrese el segundo número que sea entero: "))

    # Calcular el MCD utilizando la función
    resultado = calcular_mcd(abs(numero1), abs(numero2))

    print(f"El Máximo Común Divisor (MCD) de {numero1} y {numero2} es: {resultado}")
except ValueError:
    print("Por favor, ingrese 2 números enteros que sean tipo válidos.")
```

```python
from graphviz import Digraph

# Crear un nuevo diagrama de flujo para calcular el MCD de dos números
dot = Digraph(comment='Máximo Común Divisor (MCD)')

# Definir nodos
dot.node('A', 'Inicio', shape='oval')
dot.node('B', 'Leer A y B', shape='parallelogram')
dot.node('C', '¿B == 0?', shape='diamond')
dot.node('D', 'MCD es A', shape='box')
dot.node('E', 'Residuo = A % B', shape='box')
dot.node('F', 'A = B\nB = Residuo', shape='box')
dot.node('G', 'Fin', shape='oval')

# Definir conexiones (flechas)
dot.edge('A', 'B')
dot.edge('B', 'C')
dot.edge('C', 'D', label='Sí')
dot.edge('C', 'E', label='No')
dot.edge('E', 'F')
dot.edge('F', 'C') # Regresar a evaluar la condición
dot.edge('D', 'G')

# Mostrar el diagrama
dot
```
