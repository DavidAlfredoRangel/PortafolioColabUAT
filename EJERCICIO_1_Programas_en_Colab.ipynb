1- Un programa que convierta grados Celsius a Farenheint

> Agregar bloque entrecomillado



```python
celsius = float(input("Ingresa los grados Celsius: "))
Faren = celsius * 1.8 + 32
print(f"los grados Farenheint son {Faren}")
```

2-Un programa que convierta Kilómetros a millas.

```python
kilom = float(input("Ingresa los kilometros: "))
millas = kilom * 0.621371
print(f"los kilometros en millas son {millas}")
```

3-\Un programa que calcule la propina de una cena.

```python
cuenta = float(input("Ingresa la cantidad de la cuenta para calcular la propina de la cena "))
propina = float(input("Ahora ingresa la decimal para convertirlo en porcentaje de la propina "))
"Ejemplo 10% = 0.10, 20% = 0.20, 25% = 0.25"
total = cuenta * propina
total2 = total + cuenta

print(f"La cantidad de propina a pagar es de {total}$ y la cuenta completa es de {total2}$")

```
